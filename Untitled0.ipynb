```python
import numpy as np
import torch
import torch.nn
```

```python
v = [1,40,2,-3]

a_minimum = np.argmin(v)
a_maximum = np.argmax(v)

print("min,max: %g,%g"%(a_minimum,a_maximum))

print(f"min: {v[a_minimum]} and max: {v[a_maximum]}")
```
