```python
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
z = [1,2,3]
numerator = np.exp(z)
denominator = np.sum(np.exp(z))
sigmoid = numerator / denominator
print(sigmoid)
print(sum(sigmoid))
```

```python
x = np.random.randint(-5,16,20)
print(x)
num = np.exp(x)
den = np.sum(num)
softmax = num/den

plt.plot(x,softmax,"bo")
plt.xlabel('input')
plt.ylabel('softmax probability')

plt.title(f"$\sum\sigma$ = {np.sum(softmax)}")
```

```python
softmaxT = nn.Softmax(dim=0)
result = softmaxT(torch.Tensor(z))
print(result)
```

```python
plt.plot(sigmoid,result,'o')
plt.xlabel("manual softmax")
plt.ylabel("Pytorch nn.Softmax ")
```
